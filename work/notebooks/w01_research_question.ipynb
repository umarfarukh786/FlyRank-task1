## 2. The question: decision, action, cost of a wrong call

**Research question:** Among eligible pseudonymized content items, which pages should a FlyRank content or SEO reviewer inspect first for a possible refresh, based on observable search and engagement signals?

**Decision and action:** the reviewer receives a ranked queue, checks the reason codes and current context, then chooses `refresh`, `refresh_and_review_ctr`, `refresh_and_review_engagement`, `expand_and_refresh`, or `monitor`.

**Cost of a wrong call:** a false positive spends scarce editorial time on a page that may be healthy; a false negative delays attention to a page with meaningful visibility or engagement. The project will therefore favor a transparent baseline, Precision@K, grouped validation by client, and explicit human review over automatic changes.

```python
research_question = {
    "question": "Which existing content items should a reviewer inspect first for a possible refresh?",
    "decision_owner": "content or SEO reviewer",
    "wrong_call_cost": "editorial time wasted on false positives or delayed attention after false negatives",
    "validation_design": "client-grouped holdout when both label classes are available",
}
for label, value in research_question.items():
    print(f"{label}: {value}")
```

## 3. Quick look at the data

The starter CSV is a public-safe teaching slice: one row per pseudonymized content item with trailing-90-day activity and adjacent 30-day comparison fields. These first numbers are descriptive only; they justify investigating the lane but do not establish that any refresh will improve performance.

```python
from pathlib import Path

import pandas as pd

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent.parent
RAW_PATH = ROOT / "data" / "raw" / "content_refresh_anonymized.csv"
if not RAW_PATH.exists():
    raise FileNotFoundError(f"Starter dataset not found: {RAW_PATH}")

df = pd.read_csv(RAW_PATH)
required_columns = {
    "content_id", "client_id", "impressions_90d", "sessions_90d", "trend_direction",
}
missing_columns = required_columns.difference(df.columns)
if missing_columns:
    raise ValueError(f"Missing required columns: {sorted(missing_columns)}")

down_rate = df["trend_direction"].astype(str).str.lower().eq("down").mean()
print(f"Rows: {len(df):,}")
print(f"Pseudonymized clients: {df['client_id'].nunique():,}")
print(f"Observed down-trend rate: {down_rate:.3f}")
print(f"Median trailing-90-day impressions: {df['impressions_90d'].median():,.0f}")
print(f"Rows with measurable sessions: {(df['sessions_90d'] > 0).mean():.3f}")
assert df["content_id"].is_unique
print("Data check: one row per content item")
```
